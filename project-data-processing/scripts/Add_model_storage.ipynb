## Scripts de subida del modelo al Storage

```python
import os
from pathlib import Path
import json
from google.cloud import storage
import pickle
from dotenv import load_dotenv
```

```python
PATH_ROOT = os.getcwd()
post_project = [i for i, val in enumerate(str(PATH_ROOT).split('\\')) if val == 'project-data-processing'][0]
os.chdir('\\'.join(str(PATH_ROOT).split('\\')[:post_project + 1]))
```

```python
load_dotenv()

project_id = os.getenv("GCP_PROJECT_ID")
bucket_name = os.getenv("GCP_BUCKET_NAME")
model_ruta = os.getenv("MODEL_ROOT")
pipeline_ruta = os.getenv("PIPELINE_ROOT")
credentials_path = os.getenv("GOOGLE_APPLICATION_CREDENTIALS")

print("Proyecto:", project_id)
print("Bucket:", bucket_name)
print("Ruta del modelo:", model_ruta)
print("Ruta de la pipeline:", pipeline_ruta)
print("Credencial:", credentials_path)
```

```python
model_name  = "Model-GradientBoostingRegressor.joblib"
model_path = f"./models/{model_name}"
print(model_path)

storage_client = storage.Client(project = project_id)

destination_blob_name = f"{model_ruta}/{model_name}"
bucket = storage_client.bucket(bucket_name)

blob = bucket.blob(destination_blob_name)
blob.upload_from_filename(model_path)
```

```python
pipeline_name  = "Pipeline-Transformacion-Training.joblib"
pipeline_path = f"./models/{pipeline_name}"
print(pipeline_path)

destination_blob_name = f"{pipeline_ruta}/{pipeline_name}"
bucket = storage_client.bucket(bucket_name)

blob = bucket.blob(destination_blob_name)
blob.upload_from_filename(pipeline_path)
```

```python
metadata_name  = "metadata_transformer.py"
metadata_name_path = f"./models/{metadata_name}"
print(metadata_name_path)

destination_blob_name = f"{pipeline_ruta}/{metadata_name}"
bucket = storage_client.bucket(bucket_name)

blob = bucket.blob(destination_blob_name)
blob.upload_from_filename(metadata_name_path)
```
